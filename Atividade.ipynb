### Limitações do Uso de Apenas Uma Variável:
1. **Natureza Multifatorial da Doença:** A progressão do diabetes é influenciada por uma combinação complexa de fatores genéticos, metabólicos e hábitos de vida. Isolar apenas uma variável (como o IMC) ignora outros indicadores essenciais (como a glicemia ou pressão arterial).
2. **Baixa Capacidade Explicativa ($R^2$ baixo):** Um modelo de regressão linear simples com apenas uma variável não consegue capturar toda a variabilidade dos dados, resultando em um erro quadrático médio elevado nas previsões.
3. **Relações Não Lineares e Interações:** Uma única variável não captura o efeito combinado entre múltiplos atributos (por exemplo, a relação conjunta entre o IMC e o nível de glicose), limitando o desempenho preditivo do modelo.
